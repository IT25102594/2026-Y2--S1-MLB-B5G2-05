1. IT25101630_Data_Quality_Check




```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from imblearn.over_sampling import SMOTE

pd.set_option('display.width', 120)
```

```python
# Load the raw dataset from the group repository structure
df = pd.read_csv('data/raw/data.csv')

# Strip hidden spaces from column names
df.columns = df.columns.str.strip()

print(f"Initial Dataset Shape: {df.shape}")
```

```python
# 1. Check for exact duplicate rows
duplicate_count = df.duplicated().sum()
if duplicate_count > 0:
    df = df.drop_duplicates()
    print(f"Dropped {duplicate_count} duplicate rows.")
else:
    print("No duplicate rows found.")

# 2. Check for true missing values (NaN)
missing_total = df.isnull().sum().sum()
print(f"Total True Missing Values (NaN): {missing_total}")

# 3. Check for fake placeholders (-999 or 9999)
placeholders = (df == -999).sum().sum() + (df == -9999).sum().sum()
print(f"Total Placeholder Values (-999): {placeholders}")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Visual 1: Missing Data Heatmap
sns.heatmap(df.isnull(), yticklabels=False, cbar=False, cmap='viridis', ax=axes[0])
axes[0].set_title('Missing Data Heatmap (Solid color = No missing data)')

# Visual 2: Target Distribution Bar Chart
# FIX 1: Added hue and legend=False to silence the Seaborn warning
sns.countplot(data=df, x='Bankrupt?', hue='Bankrupt?', palette=['#4C72B0', '#C44E52'], legend=False, ax=axes[1])
axes[1].set_title('Target Distribution (Bankrupt vs Non-Bankrupt)')
axes[1].set_xlabel('Bankrupt? (0 = No, 1 = Yes)')
axes[1].set_ylabel('Count')

# FIX 2: Added ../ to the file path so it saves outside the notebooks folder
plt.tight_layout()
plt.show()
```

```python
# Separate features and target
X = df.drop(columns=['Bankrupt?'])
y = df['Bankrupt?']

# Split 80/20, using stratify to keep the rare bankruptcy cases balanced
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42)

print(f"Train Set Shape: {X_train.shape}")
print(f"Test Set Shape: {X_test.shape}")
```

2. IT25103479_Outlier_Treatment





```python
# these r the main financial ratios we want to check for outliers
key_cols = [
    'ROA(C) before interest and depreciation before interest',
    'Debt ratio %',
    'Current Ratio',
    'Net Income to Total Assets',
    'Total debt/Total net worth',
    'Revenue Per Share (Yuan ¥)',
    'Cash flow rate'
]

# plot boxplots before we fix anything
fig, axes = plt.subplots(2, 4, figsize=(18, 8))
axes = axes.flatten()

for i, c in enumerate(key_cols):
    sns.boxplot(y=X_train[c], ax=axes[i], color='#4C72B0')
    axes[i].set_title(c, fontsize=9)
    # use symlog so extreme outliers dont squash the box flat
    axes[i].set_yscale('symlog')
    axes[i].set_ylabel('')

axes[-1].axis('off')
plt.suptitle('Boxplots BEFORE Outlier Treatment (Train Data)', fontsize=13)
plt.tight_layout()
plt.show()
```

```python
# func to get iqr bounds
def get_outlier_bounds(series, k=1.5):
    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1
    lower = q1 - k * iqr
    upper = q3 + k * iqr
    return lower, upper

# create copies so we dont mess up original dfs
train_treated = X_train.copy()
test_treated = X_test.copy()
capping_log = []

# loop through each key column
for c in key_cols:
    # calc bounds using ONLY train data to stop data leakage
    lower, upper = get_outlier_bounds(X_train[c])
    before_min, before_max = X_train[c].min(), X_train[c].max()
    
    # clip both train and test sets using the train bounds
    train_treated[c] = train_treated[c].clip(lower=lower, upper=upper)
    test_treated[c] = test_treated[c].clip(lower=lower, upper=upper)
    
    capping_log.append({
        'Feature': c,
        'Min before': before_min,
        'Max before': before_max,
        'Min after': train_treated[c].min(),
        'Max after': train_treated[c].max()
    })

# show what got changed
print("--- capping log (train data) ---")
print(pd.DataFrame(capping_log))
```

```python
# plot boxplots again to show they r fixed
fig, axes = plt.subplots(2, 4, figsize=(18, 8))
axes = axes.flatten()

for i, c in enumerate(key_cols):
    sns.boxplot(y=train_treated[c], ax=axes[i], color='#55A868')
    axes[i].set_title(c, fontsize=9)
    axes[i].set_ylabel('')

axes[-1].axis('off')
plt.suptitle('Boxplots AFTER IQR Capping', fontsize=13)
plt.tight_layout()
plt.show()

# carry the treated data forward in memory (no CSV save/load)
X_train = train_treated
X_test = test_treated
print("outlier treatment applied, carried forward in memory.")
```

Member 3 — IT25100701 — Feature Scaling





```python
# setup the scaler (makes mean=0 and variance=1)
scaler = StandardScaler()

# fit ONLY on train data to prevent data leakage
scaler.fit(X_train)

# transform both train and test using train's rules
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)

# sklearn returns numpy arrays, so turn them back into pandas dfs
train_scaled_df = pd.DataFrame(X_train_scaled, columns=X_train.columns, index=X_train.index)
test_scaled_df = pd.DataFrame(X_test_scaled, columns=X_test.columns, index=X_test.index)

print("scaling done.")
```

```python
# pick one feature to show the before and after effect
feature_to_plot = 'Debt ratio %'
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# before scaling visual (X_train prior to this step)
sns.kdeplot(X_train[feature_to_plot], ax=axes[0], color='#4C72B0', fill=True)
axes[0].set_title(f'BEFORE Scaling: {feature_to_plot}')
axes[0].set_xlabel('Original Values')

# after scaling visual
sns.kdeplot(train_scaled_df[feature_to_plot], ax=axes[1], color='#55A868', fill=True)
axes[1].set_title(f'AFTER Scaling: {feature_to_plot}')
axes[1].set_xlabel('Scaled Values (Z-scores)')

plt.tight_layout()
plt.show()
```

```python
X_train_scaled = pd.DataFrame(scaler.transform(X_train), columns=X_train.columns, index=X_train.index)
X_test_scaled = pd.DataFrame(scaler.transform(X_test), columns=X_test.columns, index=X_test.index)

X_train = X_train_scaled
X_test = X_test_scaled
```

Member 4 — IT25102594 — Feature Selection





```python
# find cols where 99% or more of the data is exactly the same
quasi_constant_cols = []
for col in X_train.columns:
    # get the percentage of the most common value
    top_freq = X_train[col].value_counts(normalize=True).iloc[0]
    if top_freq >= 0.99:
        quasi_constant_cols.append(col)

# drop them from both train and test
X_train = X_train.drop(columns=quasi_constant_cols)
X_test = X_test.drop(columns=quasi_constant_cols)

print("dropped (quasi-constant):", quasi_constant_cols)
print("shape after drop:", X_train.shape)
```

```python
# Reattach target temporarily to calculate correlations on the current feature set
train_reduced = X_train.copy()
train_reduced['Bankrupt?'] = y_train

# calc correlation of all remaining features against the target
target_corr = train_reduced.corr()['Bankrupt?'].drop('Bankrupt?')

# get absolute values so we can sort them properly
target_corr_abs = target_corr.abs().sort_values(ascending=False)

# plot the top 20 most important features
plt.figure(figsize=(10, 8))
target_corr_abs.head(20).sort_values().plot(kind='barh', color='#4C72B0')
plt.title('Top 20 Features by Correlation with Bankruptcy')
plt.xlabel('Absolute Correlation Coefficient')
plt.tight_layout()
plt.show()

# drop features that have less than 0.02 correlation with target
irrelevant_cols = target_corr_abs[target_corr_abs < 0.02].index.tolist()
X_train = X_train.drop(columns=irrelevant_cols)
X_test = X_test.drop(columns=irrelevant_cols)

print(f"dropped {len(irrelevant_cols)} features with near-zero target correlation.")
print("shape after target filter:", X_train.shape)
```

```python
# get top 20 remaining features for a clean heatmap
top_features = target_corr_abs.loc[X_train.columns].head(20).index.tolist()

# calc correlation matrix just for the remaining features
corr_matrix = X_train[top_features].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=False, cmap='coolwarm', center=0, square=True)
plt.title('Multicollinearity Heatmap (Top 20 Remaining Features)')
plt.tight_layout()
plt.show()

# script to drop highly correlated feature pairs (>0.9)
corr_abs = X_train.corr().abs()
# grab just the upper triangle of the matrix to avoid checking duplicates
upper = corr_abs.where(np.triu(np.ones(corr_abs.shape), k=1).astype(bool))
to_drop = set()

for col in upper.columns:
    # find features highly correlated with this column
    correlated_with = upper.index[upper[col] > 0.9].tolist()
    for other in correlated_with:
        if col in to_drop or other in to_drop:
            continue
        # logic: keep the one that correlates better with the target
        if target_corr_abs.get(col, 0) >= target_corr_abs.get(other, 0):
            to_drop.add(other)
        else:
            to_drop.add(col)

# drop redundant cols from train and test sets
X_train = X_train.drop(columns=list(to_drop))
X_test = X_test.drop(columns=list(to_drop))

print(f"dropped {len(to_drop)} redundant features.")
print("final train shape:", X_train.shape)
```

Member 5 — IT25101761 — PCA (Dimensionality Reduction)





```python
print("starting features:", X_train.shape[1])
```

```python
# configure PCA to keep 95% of the variance
pca = PCA(n_components=0.95, random_state=42)

# FIT ONLY ON TRAIN DATA to prevent leakage
pca.fit(X_train)

# plot the cumulative explained variance for the viva
plt.figure(figsize=(8, 5))
plt.plot(np.cumsum(pca.explained_variance_ratio_), marker='o', linestyle='--', color='#4C72B0')
plt.title('PCA: Cumulative Explained Variance')
plt.xlabel('Number of Principal Components')
plt.ylabel('Cumulative Variance Retained')
plt.axhline(y=0.95, color='r', linestyle='-')
plt.text(0.5, 0.90, '95% Variance Threshold', color='red', fontsize=10)
plt.grid(True)
plt.tight_layout()
plt.show()

print(f"reduced features down to {pca.n_components_} principal components.")
```

```python
# transform both sets
X_train_pca = pca.transform(X_train)
X_test_pca = pca.transform(X_test)

# create new column names for the components (e.g., PC1, PC2)
pc_cols = [f'PC{i+1}' for i in range(X_train_pca.shape[1])]

# convert back to dataframes
train_pca_df = pd.DataFrame(X_train_pca, columns=pc_cols)
test_pca_df = pd.DataFrame(X_test_pca, columns=pc_cols)

# carry the PCA-transformed data forward in memory (no CSV save/load)
X_train = train_pca_df
X_test = test_pca_df
print("PCA transform carried forward in memory.")
```

Member 6 — IT25103651 — Class Balancing (SMOTE)





```python
print("train set shape:", X_train.shape)
print("test set shape:", X_test.shape)
```

```python
# visual 1: target distribution before smote
plt.figure(figsize=(6, 4))
sns.countplot(x=y_train, hue=y_train, palette=['#4C72B0', '#C44E52'], legend=False)
plt.title('Training Set: BEFORE SMOTE')
plt.xlabel('Bankrupt? (0 = No, 1 = Yes)')
plt.ylabel('Count')
plt.tight_layout()
plt.show()

print("before smote counts:\n", y_train.value_counts())
```

```python
# initialize SMOTE
smote = SMOTE(random_state=42)

# fit and resample ONLY the training data
X_train_smote, y_train_smote = smote.fit_resample(X_train, y_train)

print("smote applied successfully to the training data.")
```

```python
# visual 2: target distribution after smote
plt.figure(figsize=(6, 4))
sns.countplot(x=y_train_smote, hue=y_train_smote, palette=['#4C72B0', '#C44E52'], legend=False)
plt.title('Training Set: AFTER SMOTE')
plt.xlabel('Bankrupt? (0 = No, 1 = Yes)')
plt.ylabel('Count')
plt.tight_layout()
plt.show()

print("after smote counts:\n", y_train_smote.value_counts())
```

```python
# build the final dataframes for modeling
train_final = X_train_smote.copy()
train_final['Bankrupt?'] = y_train_smote

# test set remains untouched by SMOTE
test_final = X_test.copy()
test_final['Bankrupt?'] = y_test

# Save ONLY the final processed dataset (no more intermediate per-member files)
train_final.to_csv('results/outputs/final_processed_train.csv', index=False)
test_final.to_csv('results/outputs/final_processed_test.csv', index=False)

print("saved final_processed_train.csv and final_processed_test.csv to results/outputs/")
print("Final train shape:", train_final.shape)
print("Final test shape:", test_final.shape)
```
