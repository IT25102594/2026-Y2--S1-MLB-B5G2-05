```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.decomposition import PCA

pd.set_option('display.width', 120)
```

```python
# load the optimized files member 4 saved in the data folder
train_df = pd.read_csv('../data/train_selected.csv')
test_df = pd.read_csv('../data/test_selected.csv')

# separate features from the target
y_train = train_df['Bankrupt?']
X_train = train_df.drop(columns=['Bankrupt?'])

y_test = test_df['Bankrupt?']
X_test = test_df.drop(columns=['Bankrupt?'])

print("starting features:", X_train.shape[1])
```

```python
# configure PCA to keep 95% of the variance
pca = PCA(n_components=0.95, random_state=42)

# FIT ONLY ON TRAIN DATA to prevent leakage
pca.fit(X_train)

# plot the cumulative explained variance for the viva
plt.figure(figsize=(8, 5))
plt.plot(np.cumsum(pca.explained_variance_ratio_), marker='o', linestyle='--', color='#4C72B0')
plt.title('PCA: Cumulative Explained Variance')
plt.xlabel('Number of Principal Components')
plt.ylabel('Cumulative Variance Retained')
plt.axhline(y=0.95, color='r', linestyle='-')
plt.text(0.5, 0.90, '95% Variance Threshold', color='red', fontsize=10)
plt.grid(True)
plt.tight_layout()

# save visual
plt.savefig('../results/eda_visualizations/member5_pca_variance.png', dpi=110)
plt.show()

print(f"reduced features down to {pca.n_components_} principal components.")
```

```python
# transform both sets
X_train_pca = pca.transform(X_train)
X_test_pca = pca.transform(X_test)

# create new column names for the components (e.g., PC1, PC2)
pc_cols = [f'PC{i+1}' for i in range(X_train_pca.shape[1])]

# convert back to dataframes
train_pca_df = pd.DataFrame(X_train_pca, columns=pc_cols)
test_pca_df = pd.DataFrame(X_test_pca, columns=pc_cols)

# reattach the target columns
train_pca_df['Bankrupt?'] = y_train.values
test_pca_df['Bankrupt?'] = y_test.values

# save intermediate files to data folder for member 6 (SMOTE)
train_pca_df.to_csv('../data/train_pca.csv', index=False)
test_pca_df.to_csv('../data/test_pca.csv', index=False)

print("saved train_pca.csv and test_pca.csv")
```
